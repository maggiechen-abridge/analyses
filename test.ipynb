```python
from google.cloud import bigquery
import pandas as pd

# Initialize BigQuery client
client = bigquery.Client(project='abridge-client-prod')

# Test query
query = """
SELECT 
    encounter_id,
    user_id,
    recording_started_date
FROM `abridge-client-prod.dbt_prod.tblEncounter`
WHERE recording_started_date >= DATE_SUB(CURRENT_DATE(), INTERVAL 1 DAY)
LIMIT 5
"""

df = client.query(query).to_dataframe()
print(f"✅ Success! Loaded {len(df)} records")
print(df.head())
```
